```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('titanic')
df
```

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
for i in df.columns:
    if df[i].dtype == 'object'or df[i].dtype == 'category' or df[i].dtype == 'bool':
        df[i] = le.fit_transform(df[i])
```

```python
df
```

```python
df.info()
```

```python
df['age'] = df['age'].fillna(df['age'].median())
```

```python
df.info()
```

```python
z = 1
plt.figure(figsize=(12,12))
for i in df.columns:
    plt.subplot(4,4,z)
    sns.boxenplot(df[i])
    plt.xlabel(i)
    z+=1
plt.tight_layout()
plt.show()
```

```python
X = df.iloc[:,1:]
X
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
y = df.iloc[:,0]
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(X,y,random_state=1,train_size=0.7)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
y_predict = lr.predict(x_test)
```

```python
act_pred = pd.DataFrame()
act_pred['actual'] = y_test
act_pred['y_predict'] = y_predict
act_pred
```

```python
from sklearn.metrics import *
```

```python
accuracy_score(act_pred['actual'],act_pred['y_predict'])*100
```

```python
sns.regplot(x='age', y='survived', data=df, logistic=True)

plt.title("Logistic Regression: Age vs Survival")
plt.xlabel("Age")
plt.ylabel("Survival Probability")

plt.show()
```

```python
sns.regplot(x='fare',y ='survived',data= df,logistic=True)
plt.xlabel('fare')
plt.ylabel('Survived')
plt.show()
```

```python
y_pred = lr.predict(x_test)

plt.figure(figsize=(8,5))

plt.scatter(range(len(y_test)), y_test, label="Actual")
plt.scatter(range(len(y_pred)), y_pred, label="Predicted")

plt.legend()
plt.title("Actual vs Predicted Survival")

plt.show()
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")

plt.show()
```
