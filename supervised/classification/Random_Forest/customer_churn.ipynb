*** we dont need to emove the outliers in tree algo because it is robusts to outliers ***

```python
# we are dropping customer id because it dos not help in classification
# and while labelencoding its gonna use resourses 
df.drop(columns={'customerID'},inplace=True)
```

```python
df
```

```python
# Encoding - Converting catagorical val to numerical val 
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
for i in df.columns:
    if df[i].dtype == 'object': # if any column is object type convert it to numeric 
        df[i] = le.fit_transform(df[i])
```

```python
df
```

```python

x = df.drop('Churn', axis=1)
y = df['Churn']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,random_state=1,train_size=0.8)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
Model_1 = RandomForestClassifier()
```

```python
# training model 
Model_1.fit(x_train,y_train)
```

```python
y_predict = Model_1.predict(x_test)
```

```python
act_pred = pd.DataFrame()
act_pred['actual'] = y_test
act_pred['predicted'] = y_predict
act_pred
```

```python
from sklearn.metrics import *
```

```python
print(f'Accuracy of Model_1 is {accuracy_score(y_test,y_predict)*100}')
```

```python
from sklearn.ensemble import RandomForestClassifier

model_2 = RandomForestClassifier(n_estimators=200 , criterion="entropy" , max_depth=10 , min_samples_leaf=5 , min_samples_split=5)

# n_estimatorsint, default=100 .The number of trees in the forest
# criterion{“gini”, “entropy”, “log_loss”}, default=”gini”. The function to measure the quality of a split. Supported criteria are “gini” for the Gini impurity and “log_loss” and “entropy” both for the Shannon information gain
# max_depthint, default=None
# min_samples_leafint or float, default=1 . The minimum number of samples required to be at a leaf node
# min_samples_splitint or float, default=2. The minimum number of samples required to split an internal node:

# Train the model
model_2.fit(x_train , y_train)

# Ask the model to predict
y_pred = model_2.predict(x_test)

# Evalaute the performance by comparing predicted values with Actual values
from sklearn.metrics import *

print("Accuracy score of the model is :", accuracy_score(y_pred , y_test)*100)
```
