```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv(r'Boston.csv')
df
```

```python
x = df.loc[:,'CRIM':'LSTAT']
y = df.loc[:,'MEDV']
```

```python
x=np.array(x)
y=np.array(y)
```

```python
model = LinearRegression()
X_train,X_test,Y_train,Y_test = train_test_split(x,y,test_size=0.2,random_state=1) #random state is seed value it so that same random values are generated every time
model.fit(X_train,Y_train)
```

```python
m = model.coef_.reshape(-1)
c = model.intercept_.reshape(-1)
Y_predict = model.predict(X_test)
print(f'm={m}, c={c}')
```

```python
mse = mean_squared_error(Y_test, Y_predict)
print(f'Mean Squared Error: {mse}')
```
